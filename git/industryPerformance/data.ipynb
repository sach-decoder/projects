```python
df = df.drop(columns=['Company Path'])
```

```python
import sys, subprocess
subprocess.check_call([sys.executable, "-m", "pip", "install", "streamlit"])
subprocess.check_call([sys.executable, "-m", "pip", "install", "plotly"])
```

```python
#Plotting a KPI card for each company 

df_revenue_kpi = df.set_index('Company')['revenue'].to_dict()
cols = st.columns(15) 

for i in range(15):
    company = list(df_revenue_kpi.keys())[i]    # company name
    revenue = list(df_revenue_kpi.values())[i]  # revenue value
    
    cols[i].metric(
        label=f"{company} Total Revenue", 
        value=f"${revenue:,}",  
        delta="+12%", 
        border=True
    )

```

```python
# Plotting scatter plot to display operating-margin(%) vs market cap
df['operating-margin'] = df['operating-margin']*100
st.scatter_chart(df, x=df['operating-margin'], y=df['marketcap'])
```

```python
# Creating a Tree map to show the size of Industry by Revenue. 
fig = px.treemap(df, path=["Industry"], values="revenue")
st.plotly_chart(fig, use_container_width=True)
```
